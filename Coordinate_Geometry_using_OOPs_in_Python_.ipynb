# **👉Write OOP classes to handle the following scenarios**

.**👉A user can create and 2d coordinates**

.**👉A user can find out the distance between 2 coordinates**

.**👉A user can find the distance of a coordinate from origin**

.**👉A user can check if a point lines on a given line**

.**👉A user can find the distance between a given 2d point and a given line**

```python
class point:
  def __init__(self,x,y):
    self.x_cod = x
    self.y_cod = y

  def __str__(self):
    return '<{},{}>'.format(self.x_cod,self.y_cod)

  def euclidean_distance(self,other):
    return ((self.x_cod - other.x_cod)**2 + (self.y_cod - other.y_cod)**2)**0.5

  def distance_from_origin(self):
    return self.euclidean_distance(point(0,0))

class line:

  def __init__(self,A,B,C):
    self.A = A
    self.B = B
    self.C = C

  def __str__(self):
    return '{}x + {}y + {} = 0'.format(self.A,self.B,self.C)

  def point_on_line(line,point):
    if line.A*point.x_cod + line.B*point.y_cod + line.C == 0:
      return "lines on the line "
    else:
      return "does not lie on the line"



l1 = line(1,1,-2)
p1 = point(1,1)
print(l1)
print(p1)

l1.point_on_line(p1)
```

# **👉Create a class point that:**

.**👉Store x and y**

.**👉Has a method to calculate distance from origin**

.**👉Has a method to display the point**

```python
import math
class point :

  def __init__(self, x, y):
    self.x = x
    self.y = y


  def display(self):
    print(f'point:({self.x},{self.y})')


  def distance_from_origin(self):
    return math.sqrt(self.x**2 + self.y**2)


p1 =  point(3,4)
p1.display()
print("distance from origin",p1.distance_from_origin())
```

# **👉Create two point and calculate the distance between them using a class method**

```python
import math

class Point:

  def __init__(self, x, y):

    self.x = x

    self.y = y


  def distance(self, other_point):
    return math.sqrt((other_point.x - self.x)**2 + (other_point.y - self.y)**2)


p1 = Point(2, 3)
p2 = Point(6, 7)


print("Distance between points",p1.distance(p2))
```

# **👉 Build a class that checks weteher three points from a triangle ⏏**.


💡**Hint
if the area = 0 > point are collinear (no triangle)**

```python
class point:
  def __init__(self,x,y):
    self.x_cod = x
    self.y_cod = y


class triangle:
  def __init__(self,p1,p2,p3):
    self.p1 = p1
    self.p2 = p2
    self.p3 = p3


  def is_triangle(self):
    # Formula for area = 0.5 * |x1(y2 - y3) + x2(y3 - y1) + x3(y1 - y2)|
    # We only care if the expression inside the absolute value is zero for collinearity
    area_determinant = self.p1.x_cod * (self.p2.y_cod - self.p3.y_cod) + \
                       self.p2.x_cod * (self.p3.y_cod - self.p1.y_cod) + \
                       self.p3.x_cod * (self.p1.y_cod - self.p2.y_cod)


    if area_determinant == 0:
      return "not a triangle (collinear)"
    return "Triangle is possible"


# Instantiate points using the 'point' class defined in this cell
p1 = point(0,0)
p2 = point(1,2)
p3 = point(2,4)

t = triangle(p1,p2,p3)
print(t.is_triangle())

# Example of non-collinear points
p4 = point(0,0)
p5 = point(1,1)
p6 = point(0,1)
t2 = triangle(p4,p5,p6)
print(t2.is_triangle())
```
